#Returns Percentil

To clean the outlier data, it was decided to analyze temperature changes, which will be referred to as returns, as they will be treated analogously to stock returns. The `conditional_returns` function computes these returns from the dataframe after removing stuck sensor data. If the gap between observations is smaller than the threshold variable `th`, the return is calculated relative to the entry immediately preceding the gap. If the gap is larger than `th`, the return calculation is restarted.

Afterward, the distribution of returns is analyzed, and values above the 99.9th percentile and below the 0.01st percentile are discarded as outliers.


```python
from cleaning_stephanie import conditional_returns
```

```python
returns,returns_per_clean=conditional_returns(df=df_clean_stuck, th=3,low=0.001,up=0.999)
returns.head()
```

##Neighborhood Outliers

This proposal eliminates an entire neighborhood of size `segment_th` surrounding any value considered an outlier. This is because, when using the returns approach, the first point might be an outlier with respect to the previous normal trend, but subsequent points will be normal with respect to the outlier. However, only the first detected outlier will be eliminated, as subsequent returns would be normal with respect to the removed outlier.

The current method looks for outliers with a positive or negative jump and searches the surroundings for an opposite jump that adjusts to the normal value without counting intermediate NaN values ​​that might be present to remove the section completely.

#Temp All Layers

A Lasso model will be used, with the penalization parameter obtained through cross-validation using 5 consecutively divided folds. This approach takes into account the historical nature of the data, where validation should be performed on consecutive sections rather than through random splitting.

As predictor variables, the model will use humidity, radiation, and air temperature values, in addition to the complete temperature data from the previous soil layers. The target variable can be chosen either as the temperature values themselves or as the temperature changes (returns).

The returns approach was selected because it proved to be more stable and provided better performance in short terms. But when it is a large gap is better to predict temperature. In particular, returns produced smoother curves, whereas directly predicting temperatures often resulted in noisier predictions, with connections to subsequent real data points occurring through abrupt jumps.

Models allowed by the function:

*   Lasso="lasso"
*   LassoCV="lassocv"
*   Ridge="ridge"
*   Nerual Network="nn"

Returns:

*   True: Predict returns
*   False: Predict temperature

## Mixed Model
Returns and temperature will be predicted for each layer and the gap will be filled using one of the methods depending on whether the gap is greater or less than `gp_th`.

```python
from cleaning_stephanie import soil_temp_model
from cleaning_stephanie import mix_filling
```

```python
X_air_temp=df_air_temp_clean
X_air_rad=df_air_rad_clean
Y_soil_temp=df_Clean_per
Y_soil_return=returns_per_clean
Y_soil_temp_fill = Y_soil_temp.copy()
Y_soil_return_fill = Y_soil_return.copy()
```

```python
gp_th=1500

for ly in [5,20,50,100]:
    if ly==5:

        temp_ret_filled_5cm, returns_filled, model_5cm_ret=soil_temp_model(
                                      X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                      Y_soil_temp_fill, Y_soil_return_fill,
                                      returns=True,model="lassocv",graphs=False,zoom=5,
                                      window_zoom=500,depth=ly
                                      )

        temp_temp_filled_5cm, _, model_5cm_temp=soil_temp_model(
                                      X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                      Y_soil_temp_fill, Y_soil_return_fill,
                                      returns=False,model="lassocv",graphs=False,zoom=5,
                                      window_zoom=500,depth=ly
                                      )

        temp_final = mix_filling(
                                    Y_soil_temp,temp_temp_filled_5cm,
                                    temp_ret_filled_5cm,depth=ly,gap_th = gp_th)

    elif ly==20:
          temp_ret_filled_20cm, returns_filled, model_20cm_ret=soil_temp_model(
                                        X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                        Y_soil_temp_fill, Y_soil_return_fill,
                                        returns=True,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          temp_temp_filled_20cm, _, model_20cm_temp=soil_temp_model(
                                        X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                        Y_soil_temp_fill, Y_soil_return_fill,
                                        returns=False,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          temp_final = mix_filling(
                                      Y_soil_temp,temp_temp_filled_20cm,
                                      temp_ret_filled_20cm,depth=ly,gap_th = gp_th)

    elif ly==50:
          temp_ret_filled_50cm, returns_filled, model_50cm_ret=soil_temp_model(
                                        X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                        Y_soil_temp_fill, Y_soil_return_fill,
                                        returns=True,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          temp_temp_filled_50cm, _, model_50cm_temp=soil_temp_model(
                                        X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                        Y_soil_temp_fill, Y_soil_return_fill,
                                        returns=False,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          temp_final = mix_filling(
                                      Y_soil_temp,temp_temp_filled_50cm,
                                      temp_ret_filled_50cm,depth=ly,gap_th = gp_th)

    elif ly==100:
          temp_ret_filled_100cm, returns_filled, model_100cm_ret=soil_temp_model(
                                        X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                        Y_soil_temp_fill, Y_soil_return_fill,
                                        returns=True,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          temp_temp_filled_100cm, _, model_100cm_temp=soil_temp_model(
                                        X_air_temp,X_air_rad,Y_soil_temp,Y_soil_return,
                                        Y_soil_temp_fill, Y_soil_return_fill,
                                        returns=False,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          temp_final = mix_filling(
                                      Y_soil_temp,temp_temp_filled_100cm,
                                      temp_ret_filled_100cm,depth=ly,gap_th = gp_th)

    # nombre columna
    col = f"soil_temp_deg_C_at_{ly}_cm_depth"

    # actualizar temperaturas
    Y_soil_temp_fill[col] = temp_final

    # actualizar returns
    Y_soil_return_fill[col] = returns_filled
```

Returns and moisture will be predicted for each layer and the gap will be filled using one of the methods depending on whether the gap is greater or less than `gp_th`.

```python
gp_th_5=100
gp_th_20=100
gp_th_50=100
gp_th_100=200

for ly in [5,20,50,100]:
    if ly==5:

        mois_ret_filled_5cm, returns_filled, model_5cm_ret=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                      returns=True,model="lassocv",graphs=False,zoom=5,
                                      window_zoom=500,depth=ly
                                      )

        mois_mois_filled_5cm, _, model_5cm_temp=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                      returns=False,model="lassocv",graphs=False,zoom=5,
                                      window_zoom=500,depth=ly
                                      )

        mois_final = mix_filling_moisture(
                                    Y_soil_moist,mois_mois_filled_5cm,
                                    mois_ret_filled_5cm,depth=ly,gap_th = gp_th_5)

    elif ly==20:
          mois_ret_filled_20cm, returns_filled, model_20cm_ret=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                        returns=True,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          mois_mois_filled_20cm, _, model_20cm_temp=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                        returns=False,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          mois_final = mix_filling_moisture(
                                      Y_soil_moist,mois_mois_filled_20cm,
                                      mois_ret_filled_20cm,depth=ly,gap_th = gp_th_20)

    elif ly==50:
          mois_ret_filled_50cm, returns_filled, model_50cm_ret=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                        returns=True,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          mois_mois_filled_50cm, _, model_50cm_temp=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                        returns=False,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          mois_final = mix_filling_moisture(
                                      Y_soil_moist,mois_mois_filled_50cm,
                                      mois_ret_filled_50cm,depth=ly,gap_th = gp_th_50)

    elif ly==100:
          mois_ret_filled_100cm, returns_filled, model_100cm_ret=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                        returns=True,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          mois_mois_filled_100cm, _, model_100cm_temp=soil_moist_model(
                                        X_air_temp,X_air_rad, Y_soil_ret, Y_soil_return_fill,
                                        Y_soil_moist, Y_soil_moist_return,
                                        Y_soil_moist_fill, Y_soil_moist_return_fill,
                                        returns=False,model="lassocv",graphs=False,zoom=5,
                                        window_zoom=500,depth=ly
                                        )

          mois_final = mix_filling_moisture(
                                      Y_soil_moist,mois_mois_filled_100cm,
                                      mois_ret_filled_100cm,depth=ly,gap_th = gp_th_100)

    # nombre columna
    col = f"soil_moisture_percent_at_{ly}_cm_depth"

    # actualizar temperaturas
    Y_soil_moist_fill[col] = mois_final

    # actualizar returns
    Y_soil_moist_return_fill[col] = returns_filled
```
